## Tuples

A Tuple is a collection which is ordered and unchangeable. Tuples are written with round brackets. The values can be repeated, but their number is always finite.


```python
# Is this a tuple?
t1 = ('a')

type(t1)

```

```python
# Is this a tuple
t2 = ("b", )

type(t2)
```
